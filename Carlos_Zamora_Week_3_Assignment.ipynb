```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Data
study_hours = [1, 2, 2.5, 3, 4, 5, 5.5, 6, 7, 8]
exam_scores = [55, 60, 63, 66, 72, 78, 80, 84, 88, 94]
```

```python
# Part 1: Scatter Plot
### Task 1: Study Time vs. Exam Score

#### Create Scatter Plot
plt.scatter(study_hours, exam_scores, marker = 'o', color = "red")
plt.title("Hours Studied vs. Exam Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Scores")

plt.show()
```

**What relationship do you observe between study time and exam score?**


When observing the relationship between study time and exam scores, there seems to be a linear relationship between the two. This means that while the hours studied increases, the exam scores earned also tend to increase. On the other hand, if hours studied decreases, then Exam scores also decrease.

```python
from matplotlib.lines import lineMarkers
# Part 2: Line Plot
## Task 2: Temperature Over One Week

#### Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
city_a = [68, 70, 72, 71, 74, 76, 75]
city_b = [62, 64, 67, 66, 69, 71, 70]

#### Create Line Plot
plt.plot(days, city_a, label = "City A", linestyle = 'solid')
plt.plot(days, city_b, label = "City B", linestyle = 'dashed')

plt.title("Temperature Over One Week: City A vs. City B")
plt.xlabel("Days Of The Week")
plt.ylabel("Temperature")
plt.legend()

plt.show()
```

**Which city was generally warmer during the week?**

Throughout the week, City A was generally Warmer than City B. We see the trend for both cities, and City A's trend lies above City B's trend, meaning higher temperature during the week.

```python
# Part 3: Bar Graphs - Crime Data Visualization(2022,2023)
plt.style.use("ggplot") # Sets the grid
# Data
offense = ["Aggravated Assault", "Burglary", "Larceny", "Motor Vehicle Theft", "Murder", "Rape", "Robbery"]
offenses_per100k = [264.1, 250.7, 1347.2, 318.7, 5.7, 38.0, 66.5]
offenses_change = [-3.3, -8.1, -4.9, +12.0, -12.0, -9.8, -0.8]

### Task 3A
#### Create Bar Plot
plt.figure(figsize = (15,10)) # Made Labels more readable

plt.bar(offense, offenses_per100k, color = "red")
plt.xlabel("Crimes")
plt.ylabel("Offenses Per 100 Thousand People")
plt.title("Crimes Rates Per 100,000 People in 2023")


plt.show()
```

```python
### Task 3B

colors = ["red", "blue", "green", "purple", "yellow", "pink", "black"]
plt.barh(offense, offenses_change, color = colors)
plt.title("Changes In Crime Rates (2022-2023)")
plt.xlabel("Change In offenses per 100,000 People")
plt.ylabel("Crime")
```

**Why might this graph be more useful than the graph in Task 3A**

**Think about the difference between looking at the size of the crime rate in one year and looking at how much the rate changed from one year to the next.**

This graph might be more useful than the graph in Tast 3A because seeing the overall changes from year to year, in this case year 2022-2023 is more informational on how crime rates have evolved, rather than only looking at crime rates for one year(2023). Being able to compared and see the crime rates changes can bring more conclusions to the data, and many more questions can be asked. Graph in part 3A is more like a summary of crimes, while graph 3B demonstrates the changes of crime rates over time.





```python
### Task 3C: Improve the clarity of your bar graph

#### Removed colors as multiple colors can be quite distracting

#### rearanged offense, and offenses_changes for ordering of categories

offense_sorted = ["Motor Vehicle Theft","Murder", "Rape", "Burglary", "Larceny", "Aggravated Assault","Robbery"]
offenses_change_sorted = [+12.0, -12.0, -9.8, -8.1, -4.9, -3.3, -0.8]

#### Updated plot
plt.figure(figsize= (10,5)) # better clarity
plt.barh(offense_sorted, offenses_change_sorted, color = 'dodgerblue')
plt.title("Changes In Crime Rates (2022-2023)")
plt.xlabel("Change In offenses per 100,000 People")
plt.ylabel("Crime")
plt.xticks([-15,-10,-5,0,5,10,15]) # better x-axis range


plt.tight_layout()
plt.show()
```

```python
### Task 3D: Multi-Series Grouped Bar Chart
companies = ["Company A", "Company B", "Company C"]
product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]

#### Generate indexes for the x-axis - equal to number of company categories (Company A,B,C)
x_indexes = np.arange(len(companies))
print(x_indexes)

plt.figure(figsize= (10,5))
#### creates the three bars for each company, separated correctly
plt.bar(x_indexes - 0.250, product1, width = 0.25, label = 'Product 1', color = 'red')
plt.bar(x_indexes, product2, width = 0.25, label = 'Product 2', color = 'blue')
plt.bar(x_indexes + 0.250, product3, width = 0.25, label = 'Product 3', color = 'darkgreen')

plt.title('Companies vs. Product Sales')
plt.xlabel('Companies')
plt.ylabel('Product Sales')
#### Labels the companies on the x-axis
plt.xticks(ticks = x_indexes, labels = companies)

plt.legend()
plt.show()
```
